### Dependencies & Notebook Configurations

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay, f1_score, precision_score, recall_score
from sklearn.model_selection import train_test_split
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset

# local libraries
from utils import train_and_eval, test

from plot import (
    plot_losses,
    plot_accuracies,
    plot_f1s,
    plot_precisions_and_recalls,
    plot_confusion_matrix
)
```

```python
# for reproducibility
np.random.seed(0)
torch.manual_seed(0)
```

```python
# Use GPU if available
if torch.cuda.is_available():
    DEVICE = torch.device("cuda")
else:
    DEVICE = torch.device("cpu")

USE_CUDA = DEVICE.type  == "cuda"

print(f"DEVICE: {DEVICE}")
print(f"USE_CUDA: {USE_CUDA}")
```

## Data Load & Preprocessing

```python
# Set A
SOURCE_A_TRAIN = pd.read_csv("./dataset/dataset_a_train.csv")
SOURCE_A_TEST = pd.read_csv("./dataset/dataset_a_test.csv")

display(SOURCE_A_TRAIN.head())
display(SOURCE_A_TEST.head())

# Set B
SOURCE_B_TEST = pd.read_csv("./dataset/dataset_b_test.csv")

display(SOURCE_B_TEST.head())

# Set C
SOURCE_C_TRAIN = pd.read_csv("./dataset/dataset_c_train.csv")
SOURCE_C_TEST = pd.read_csv("./dataset/dataset_c_test.csv")

display(SOURCE_C_TRAIN.head())
display(SOURCE_C_TEST.head())
```

```python
LABELS = pd.DataFrame({
    "Label": range(1, 6),
    "Set A": ["Politics", "Sports", "Business", "Technology", "Entertainment"],
    "Set B": ["Politics", "Sports", "Business", "Technology", "Entertainment"],
    "Set C": ["World", "Sports", "Business", "Technology", None]
}).set_index("Label")

display(LABELS)
```

```python
LABELS_A_NUNIQUE = LABELS["Set A"].nunique()
LABELS_B_NUNIQUE = LABELS["Set B"].nunique()
LABELS_C_NUNIQUE = LABELS["Set C"].nunique()
```

## Test Different Text Representations

```python
# Set A
a_train, a_validation = train_test_split(
    SOURCE_A_TRAIN,
    test_size=0.2,
    random_state=0,
    stratify=SOURCE_A_TRAIN["Label"]  # label-stratified split
)

print(SOURCE_A_TRAIN.shape)
display(a_train)
display(a_validation)
```

```python
BATCH_SIZE = 32
```

```python
vectorizer = TfidfVectorizer()  # default ngram_range is (1, 1), i.e. unigram

a_train_title_tfidf = vectorizer.fit_transform(a_train["Title"])
a_validation_title_tfidf = vectorizer.transform(a_validation["Title"])

a_train_title_tensor = torch.tensor(a_train_title_tfidf.toarray(), dtype=torch.float32).to(DEVICE)
a_validation_title_tensor = torch.tensor(a_validation_title_tfidf.toarray(), dtype=torch.float32).to(DEVICE)
print(a_train_title_tensor)
print(a_validation_title_tensor)

# CrossEntropyLoss expects zero-indexed labels, so label - 1
a_train_labels_tensor = torch.tensor((a_train["Label"] - 1).to_list(), dtype=torch.long).to(DEVICE)
a_validation_labels_tensor = torch.tensor((a_validation["Label"] - 1).to_list(), dtype=torch.long).to(DEVICE)
print(a_train_labels_tensor)
print(a_validation_labels_tensor)
```

```python
# Set A
a_train_title_dataset = TensorDataset(
    a_train_title_tensor.cpu(),
    a_train_labels_tensor.cpu()
)
a_validation_title_dataset = TensorDataset(
    a_validation_title_tensor.cpu(),
    a_validation_labels_tensor.cpu()
)

a_train_title_loader = DataLoader(
    a_train_title_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,  # ensure the training data is shuffled in every epoch
    pin_memory=USE_CUDA
)
a_validation_title_loader = DataLoader(
    a_validation_title_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    pin_memory=USE_CUDA
)
```

```python
class CustomMLP(nn.Module):
    def __init__(self, input_dim, output_dim):
        super().__init__()

        self.layers = nn.Sequential(
            nn.Linear(input_dim, 128),
            nn.ReLU(),
            nn.Dropout(0.1),

            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Dropout(0.4),

            nn.Linear(64, output_dim)
        )

    def forward(self, x):
        return self.layers(x)
```

```python
# Constants & Hyperparameters
MAX_EPOCH = 100
EPOCHS = range(1, MAX_EPOCH + 1)

LEARNING_RATE = 0.006
MOMENTUM = 0.9
WEIGHT_DECAY = 0.0008
```

### Model using Title Field Only

```python
model_title_only = CustomMLP(a_train_title_tensor.shape[1], LABELS_A_NUNIQUE).to(DEVICE)

criterion_title_only = nn.CrossEntropyLoss()
optimizer_title_only = optim.SGD(model_title_only.parameters(), lr=LEARNING_RATE, momentum=MOMENTUM, weight_decay=WEIGHT_DECAY)
```

```python
(
    training_losses,
    validation_losses,

    training_accuracies,
    validation_accuracies,

    training_f1macros,
    training_f1s,
    training_precisions,
    training_recalls,

    validation_f1macros,
    validation_f1s,
    validation_precisions,
    validation_recalls,

    training_trues,
    training_predictions,

    validation_trues,
    validation_predictions
) = train_and_eval(
    model_title_only,
    criterion_title_only,
    optimizer_title_only,
    MAX_EPOCH,
    a_train_title_loader,
    a_train_title_dataset,
    a_validation_title_loader,
    a_validation_title_dataset,
    device=DEVICE
)
```

```python
model_name = "MLP w/ Title-only"
hyperparams = {
    "learning_rate": LEARNING_RATE,
    "batch_size": BATCH_SIZE,
    "momentum": MOMENTUM,
    "weight_decay": WEIGHT_DECAY
}

plot_losses(
    EPOCHS,
    training_losses,
    validation_losses,
    model_name,
    hyperparams)

plot_accuracies(
    EPOCHS,
    training_accuracies,
    validation_accuracies,
    model_name,
    hyperparams)

plot_f1s(
    EPOCHS,
    training_f1macros,
    validation_f1macros,
    training_f1s,
    validation_f1s,
    model_name,
    hyperparams)

plot_precisions_and_recalls(
    EPOCHS,
    training_precisions,
    validation_precisions,
    training_recalls,
    validation_recalls,
    model_name,
    hyperparams)

plot_confusion_matrix(
    LABELS["Set A"].to_list(),
    validation_trues,
    validation_predictions,
    model_name,
    hyperparams)
```

## Refit & Test

```python
a_test_tfidf = vectorizer.transform(SOURCE_A_TEST["Title"])

a_test_tensor = torch.tensor(a_test_tfidf.toarray(), dtype=torch.float32).to(DEVICE)
print(a_test_tensor)

a_test_labels_tensor = torch.tensor((SOURCE_A_TEST["Label"] - 1).to_list(), dtype=torch.long).to(DEVICE)
print(a_test_labels_tensor)
```

```python
a_test_dataset = TensorDataset(
    a_test_tensor.cpu(),
    a_test_labels_tensor.cpu()
)

a_test_loader = DataLoader(
    a_test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    pin_memory=USE_CUDA
)
```

```python
(
    testing_predictions,
    testing_trues,
    misclassified_predictions
) = test(
    model_title_only,
    a_test_loader,
    device=DEVICE
)
```

## Combined Unigram-Bigram TF-IDF Representation

```python
bigram_vectorizer = TfidfVectorizer(ngram_range=(1, 2))

def concat_article(row):
    return row["Title"] + ' ' + row["Description"]

a_train_tfidf_bigram = bigram_vectorizer.fit_transform(a_train.apply(concat_article, axis=1))
a_validation_tfidf_bigram = bigram_vectorizer.transform(a_validation.apply(concat_article, axis=1))

a_train_bigram_tensor = torch.tensor(a_train_tfidf_bigram.toarray(), dtype=torch.float32).to(DEVICE)
a_validation_bigram_tensor = torch.tensor(a_validation_tfidf_bigram.toarray(), dtype=torch.float32).to(DEVICE)
```

```python
a_train_bigram_dataset = TensorDataset(
    a_train_bigram_tensor.cpu(),
    a_train_labels_tensor.cpu()
)
a_validation_bigram_dataset = TensorDataset(
    a_validation_bigram_tensor.cpu(),
    a_validation_labels_tensor.cpu()
)

a_train_bigram_loader = DataLoader(
    a_train_bigram_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    pin_memory=USE_CUDA
)
a_validation_bigram_loader = DataLoader(
    a_validation_bigram_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    pin_memory=USE_CUDA
)
```

```python
model_bigram = CustomMLP(a_train_bigram_tensor.shape[1], LABELS_A_NUNIQUE).to(DEVICE)

criterion_bigram = nn.CrossEntropyLoss()
optimizer_bigram = optim.SGD(model_bigram.parameters(), lr=LEARNING_RATE, momentum=MOMENTUM, weight_decay=WEIGHT_DECAY)
```

```python
(
    training_losses,
    validation_losses,

    training_accuracies,
    validation_accuracies,

    training_f1macros,
    training_f1s,
    training_precisions,
    training_recalls,

    validation_f1macros,
    validation_f1s,
    validation_precisions,
    validation_recalls,

    training_trues,
    training_predictions,

    validation_trues,
    validation_predictions
) = train_and_eval(
    model_bigram,
    criterion_bigram,
    optimizer_bigram,
    MAX_EPOCH,
    a_train_bigram_loader,
    a_train_bigram_dataset,
    a_validation_bigram_loader,
    a_validation_bigram_dataset,
    device=DEVICE
)
```

```python
model_name = "MLP w/ Unigram-Bigram TF-IDF"
hyperparams = {
    "learning_rate": LEARNING_RATE,
    "batch_size": BATCH_SIZE,
    "momentum": MOMENTUM,
    "weight_decay": WEIGHT_DECAY
}

plot_losses(
    EPOCHS,
    training_losses,
    validation_losses,
    model_name,
    hyperparams)

plot_accuracies(
    EPOCHS,
    training_accuracies,
    validation_accuracies,
    model_name,
    hyperparams)

plot_f1s(
    EPOCHS,
    training_f1macros,
    validation_f1macros,
    training_f1s,
    validation_f1s,
    model_name,
    hyperparams)

plot_precisions_and_recalls(
    EPOCHS,
    training_precisions,
    validation_precisions,
    training_recalls,
    validation_recalls,
    model_name,
    hyperparams)

plot_confusion_matrix(
    LABELS["Set A"].to_list(),
    validation_trues,
    validation_predictions,
    model_name,
    hyperparams)
```
