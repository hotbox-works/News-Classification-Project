### Dependencies & Notebook Configurations

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay, f1_score, precision_score, recall_score
from sklearn.model_selection import train_test_split
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
```

```python
# for reproducibility
np.random.seed(0)
torch.manual_seed(0)
```

```python
# Use GPU if available
if torch.cuda.is_available():
    DEVICE = torch.device("cuda")
else:
    DEVICE = torch.device("cpu")

USE_CUDA = DEVICE.type  == "cuda"

print(f"DEVICE: {DEVICE}")
print(f"USE_CUDA: {USE_CUDA}")
```

## Data Load & Preprocessing

```python
# Set A
SOURCE_A_TRAIN = pd.read_csv("./dataset/dataset_a_train.csv")
SOURCE_A_TEST = pd.read_csv("./dataset/dataset_a_test.csv")

display(SOURCE_A_TRAIN.head())
display(SOURCE_A_TEST.head())

# Set B
SOURCE_B_TEST = pd.read_csv("./dataset/dataset_b_test.csv")

display(SOURCE_B_TEST.head())

# Set C
SOURCE_C_TRAIN = pd.read_csv("./dataset/dataset_c_train.csv")
SOURCE_C_TEST = pd.read_csv("./dataset/dataset_c_test.csv")

display(SOURCE_C_TRAIN.head())
display(SOURCE_C_TEST.head())
```

```python
LABELS = pd.DataFrame({
    "Label": range(1, 6),
    "Set A": ["Politics", "Sports", "Business", "Technology", "Entertainment"],
    "Set B": ["Politics", "Sports", "Business", "Technology", "Entertainment"],
    "Set C": ["World", "Sports", "Business", "Technology", None]
}).set_index("Label")

display(LABELS)
```

```python
LABELS_A_NUNIQUE = LABELS["Set A"].nunique()
LABELS_B_NUNIQUE = LABELS["Set B"].nunique()
LABELS_C_NUNIQUE = LABELS["Set C"].nunique()
```

## Test Different Text Representations

```python
# Set A
a_train, a_validation = train_test_split(
    SOURCE_A_TRAIN,
    test_size=0.2,
    random_state=0,
    stratify=SOURCE_A_TRAIN["Label"]  # label-stratified split
)

print(SOURCE_A_TRAIN.shape)
display(a_train)
display(a_validation)
```

```python
BATCH_SIZE = 32
```

```python
vectorizer = TfidfVectorizer()  # default ngram_range is (1, 1), i.e. unigram

a_train_title_tfidf = vectorizer.fit_transform(a_train["Title"])
a_validation_title_tfidf = vectorizer.transform(a_validation["Title"])

a_train_title_tensor = torch.tensor(a_train_title_tfidf.toarray(), dtype=torch.float32).to(DEVICE)
a_validation_title_tensor = torch.tensor(a_validation_title_tfidf.toarray(), dtype=torch.float32).to(DEVICE)
print(a_train_title_tensor)
print(a_validation_title_tensor)

# CrossEntropyLoss expects zero-indexed labels, so label - 1
a_train_labels_tensor = torch.tensor((a_train["Label"] - 1).to_list(), dtype=torch.long).to(DEVICE)
a_validation_labels_tensor = torch.tensor((a_validation["Label"] - 1).to_list(), dtype=torch.long).to(DEVICE)
print(a_train_labels_tensor)
print(a_validation_labels_tensor)
```

```python
# Set A
a_train_title_dataset = TensorDataset(
    a_train_title_tensor.cpu(),
    a_train_labels_tensor.cpu()
)
a_validation_title_dataset = TensorDataset(
    a_validation_title_tensor.cpu(),
    a_validation_labels_tensor.cpu()
)

a_train_title_loader = DataLoader(
    a_train_title_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,  # ensure the training data is shuffled in every epoch
    pin_memory=USE_CUDA
)
a_validation_title_loader = DataLoader(
    a_validation_title_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    pin_memory=USE_CUDA
)
```

```python
class CustomMLP(nn.Module):
    def __init__(self, input_dim, output_dim):
        super().__init__()

        self.layers = nn.Sequential(
            nn.Linear(input_dim, 128),
            nn.ReLU(),
            nn.Dropout(0.1),

            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Dropout(0.4),

            nn.Linear(64, output_dim)
        )

    def forward(self, x):
        return self.layers(x)
```

```python
# Constants & Hyperparameters
MAX_EPOCH = 35
EPOCHS = range(1, MAX_EPOCH + 1)

LEARNING_RATE = 0.021
MOMENTUM = 0.9
WEIGHT_DECAY = 0.0017
```

### Model using Title Field Only

```python
model = CustomMLP(a_train_title_tensor.shape[1], LABELS_A_NUNIQUE).to(DEVICE)

criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(model.parameters(), lr=LEARNING_RATE, momentum=MOMENTUM, weight_decay=WEIGHT_DECAY)
```

```python
training_losses = []
validation_losses = []

training_accuracies = []
validation_accuracies = []

training_f1macros = []
training_f1s = []
training_precisions = []
training_recalls = []

validation_f1macros = []
validation_f1s = []
validation_precisions = []
validation_recalls = []

for epoch in range(MAX_EPOCH):
    print(f"Epoch: {epoch + 1:02}/{MAX_EPOCH}", end='')
    # Training
    model.train()

    training_loss = 0  # compute the cumulative loss within an epoch
    training_accuracy = 0

    training_predictions = []
    training_trues = []

    for i, (x, y) in enumerate(a_train_title_loader):
        x = x.to(DEVICE, non_blocking=True)
        y = y.to(DEVICE, non_blocking=True)

        logits = model(x)

        loss = criterion(logits, y)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        # multiply by batch size to get the total loss for this batch
        training_loss += loss.item() * x.size(0)

        predictions = torch.argmax(logits, dim=1)

        training_predictions.extend(predictions.detach().cpu().numpy())
        training_trues.extend(y.cpu().numpy())

        training_accuracy += (predictions == y).sum().item()

    training_loss /= len(a_train_title_dataset)
    training_accuracy /= len(a_train_title_dataset)

    training_losses.append(training_loss)
    training_accuracies.append(training_accuracy)

    training_f1macro = f1_score(training_trues, training_predictions, average="macro", zero_division=0)
    training_f1macros.append(training_f1macro)

    training_f1 = f1_score(training_trues, training_predictions, average="weighted", zero_division=0)
    training_f1s.append(training_f1)

    training_precision = precision_score(training_trues, training_predictions, average="weighted", zero_division=0)
    training_precisions.append(training_precision)

    training_recall = recall_score(training_trues, training_predictions, average="weighted", zero_division=0)
    training_recalls.append(training_recall)

    # Validation
    model.eval()

    validation_loss = 0
    validation_accuracy = 0

    validation_predictions = []
    validation_trues = []

    with torch.inference_mode():
        for i, (x, y) in enumerate(a_validation_title_loader):
            x = x.to(DEVICE, non_blocking=True)
            y = y.to(DEVICE, non_blocking=True)

            logits = model(x)

            loss = criterion(logits, y)

            validation_loss += loss.item() * x.size(0)

            predictions = torch.argmax(logits, dim=1)

            validation_predictions.extend(predictions.detach().cpu().numpy())
            validation_trues.extend(y.cpu().numpy())

            validation_accuracy += (predictions == y).sum().item()

    validation_loss /= len(a_validation_title_dataset)
    validation_accuracy /= len(a_validation_title_dataset)

    validation_losses.append(validation_loss)
    validation_accuracies.append(validation_accuracy)

    validation_f1macro = f1_score(validation_trues, validation_predictions, average="macro", zero_division=0)
    validation_f1macros.append(validation_f1macro)

    validation_f1 = f1_score(validation_trues, validation_predictions, average="weighted", zero_division=0)
    validation_f1s.append(validation_f1)

    validation_precision = precision_score(validation_trues, validation_predictions, average="weighted", zero_division=0)
    validation_precisions.append(validation_precision)

    validation_recall = recall_score(validation_trues, validation_predictions, average="weighted", zero_division=0)
    validation_recalls.append(validation_recall)

    print("\tDone!")
```

```python
plt.figure(figsize=(10, 7))

plt.plot(EPOCHS, training_losses, label="Training Loss")
plt.plot(EPOCHS, validation_losses, label="Validation Loss")

plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.title("Training & Validation Losses (Title)")

plt.figtext(1, 0, f"Learning Rate: {LEARNING_RATE} / Batch Size: {BATCH_SIZE} / Momentum: {MOMENTUM} / Weight Decay: {WEIGHT_DECAY}", ha="right")

plt.legend()

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 7))

plt.plot(EPOCHS, training_accuracies, label="Training Accuracy")
plt.plot(EPOCHS, validation_accuracies, label="Validation Accuracy")

plt.xlabel("Epoch")
plt.ylabel("Accuracy")

plt.title("Training & Validation Accuracies (Title)")

plt.figtext(1, 0, f"Learning Rate: {LEARNING_RATE} / Batch Size: {BATCH_SIZE} / Momentum: {MOMENTUM} / Weight Deacy: {WEIGHT_DECAY}", ha="right")

plt.legend()

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 7))

plt.plot(
    EPOCHS,
    training_f1macros,
    label="Training Macro-F1"
)

plt.plot(
    EPOCHS,
    validation_f1macros,
    label="Validation Macro-F1"
)

plt.plot(
    EPOCHS,
    training_f1s,
    label="Training F1 (Weighted)"
)

plt.plot(
    EPOCHS,
    validation_f1s,
    label="Validation F1 (Weighted)"
)

plt.xlabel("Epoch")
plt.ylabel("F1 Score")

plt.title("Training & Validation Macro-F1/F1 Scores (Title)")

plt.figtext(1, 0, f"Learning Rate: {LEARNING_RATE} / Batch Size: {BATCH_SIZE} / Momentum: {MOMENTUM} / Weight Deacy: {WEIGHT_DECAY}", ha="right")

plt.legend()

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 7))

plt.plot(
    EPOCHS,
    training_precisions,
    label="Training Precision"
)

plt.plot(
    EPOCHS,
    validation_precisions,
    label="Validation Precision"
)

plt.plot(
    EPOCHS,
    training_recalls,
    label="Training Recall"
)

plt.plot(
    EPOCHS,
    validation_recalls,
    label="Validation Recall"
)

plt.xlabel("Epoch")
plt.ylabel("Score")
plt.title("Training & Validation Precision/Recall (Title)")

plt.figtext(1, 0, f"Learning Rate: {LEARNING_RATE} / Batch Size: {BATCH_SIZE} / Momentum: {MOMENTUM} / Weight Deacy: {WEIGHT_DECAY}", ha="right")

plt.legend()

plt.tight_layout()
plt.show()
```

```python
cm = confusion_matrix(
    validation_trues,
    validation_predictions
)

cm_display = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=LABELS["Set A"].to_list())  # overriding display_labels to show the actual labels
cm_display.plot(values_format='d')

plt.title("Validation Confusion Matrix (Title)")

plt.figtext(1, 0, f"Learning Rate: {LEARNING_RATE} / Batch Size: {BATCH_SIZE} / Momentum: {MOMENTUM} / Weight Deacy: {WEIGHT_DECAY}", ha="right")

plt.tight_layout()
plt.show()
```

## Combined Unigram-Bigram TF-IDF Representation

```python
bigram_vectorizer = TfidfVectorizer(ngram_range=(1, 2))

def concat_article(row):
    return row["Title"] + ' ' + row["Description"]

a_train_tfidf_bigram = bigram_vectorizer.fit_transform(a_train.apply(concat_article, axis=1))
a_validation_tfidf_bigram = bigram_vectorizer.transform(a_validation.apply(concat_article, axis=1))

a_train_bigram_tensor = torch.tensor(a_train_tfidf_bigram.toarray(), dtype=torch.float32).to(DEVICE)
a_validation_bigram_tensor = torch.tensor(a_validation_tfidf_bigram.toarray(), dtype=torch.float32).to(DEVICE)
```

```python
a_train_bigram_dataset = TensorDataset(
    a_train_bigram_tensor.cpu(),
    a_train_labels_tensor.cpu()
)
a_validation_bigram_dataset = TensorDataset(
    a_validation_bigram_tensor.cpu(),
    a_validation_labels_tensor.cpu()
)

a_train_bigram_loader = DataLoader(
    a_train_bigram_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    pin_memory=USE_CUDA
)
a_validation_bigram_loader = DataLoader(
    a_validation_bigram_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    pin_memory=USE_CUDA
)
```

```python
bigram_model = CustomMLP(a_train_bigram_tensor.shape[1], LABELS_A_NUNIQUE).to(DEVICE)

criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(bigram_model.parameters(), lr=LEARNING_RATE, momentum=MOMENTUM, weight_decay=WEIGHT_DECAY)
```

```python
training_losses = []
validation_losses = []

training_accuracies = []
validation_accuracies = []

training_f1macros = []
training_f1s = []
training_precisions = []
training_recalls = []

validation_f1macros = []
validation_f1s = []
validation_precisions = []
validation_recalls = []

for epoch in range(MAX_EPOCH):
    print(f"Epoch: {epoch + 1:02}/{MAX_EPOCH}", end='')
    # Training
    bigram_model.train()

    training_loss = 0  # compute the cumulative loss within an epoch
    training_accuracy = 0

    training_predictions = []
    training_trues = []

    for i, (x, y) in enumerate(a_train_bigram_loader):
        x = x.to(DEVICE, non_blocking=True)
        y = y.to(DEVICE, non_blocking=True)

        logits = bigram_model(x)

        loss = criterion(logits, y)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        # multiply by batch size to get the total loss for this batch
        training_loss += loss.item() * x.size(0)

        predictions = torch.argmax(logits, dim=1)

        training_predictions.extend(predictions.detach().cpu().numpy())
        training_trues.extend(y.cpu().numpy())

        training_accuracy += (predictions == y).sum().item()

    training_loss /= len(a_train_bigram_dataset)
    training_accuracy /= len(a_train_bigram_dataset)

    training_losses.append(training_loss)
    training_accuracies.append(training_accuracy)

    training_f1macro = f1_score(training_trues, training_predictions, average="macro", zero_division=0)
    training_f1macros.append(training_f1macro)

    training_f1 = f1_score(training_trues, training_predictions, average="weighted", zero_division=0)
    training_f1s.append(training_f1)

    training_precision = precision_score(training_trues, training_predictions, average="weighted", zero_division=0)
    training_precisions.append(training_precision)

    training_recall = recall_score(training_trues, training_predictions, average="weighted", zero_division=0)
    training_recalls.append(training_recall)

    # Validation
    bigram_model.eval()

    validation_loss = 0
    validation_accuracy = 0

    validation_predictions = []
    validation_trues = []

    with torch.inference_mode():
        for i, (x, y) in enumerate(a_validation_bigram_loader):
            x = x.to(DEVICE, non_blocking=True)
            y = y.to(DEVICE, non_blocking=True)

            logits = bigram_model(x)

            loss = criterion(logits, y)

            validation_loss += loss.item() * x.size(0)

            predictions = torch.argmax(logits, dim=1)

            validation_predictions.extend(predictions.detach().cpu().numpy())
            validation_trues.extend(y.cpu().numpy())

            validation_accuracy += (predictions == y).sum().item()

    validation_loss /= len(a_validation_bigram_dataset)
    validation_accuracy /= len(a_validation_bigram_dataset)

    validation_losses.append(validation_loss)
    validation_accuracies.append(validation_accuracy)

    validation_f1macro = f1_score(validation_trues, validation_predictions, average="macro", zero_division=0)
    validation_f1macros.append(validation_f1macro)

    validation_f1 = f1_score(validation_trues, validation_predictions, average="weighted", zero_division=0)
    validation_f1s.append(validation_f1)

    validation_precision = precision_score(validation_trues, validation_predictions, average="weighted", zero_division=0)
    validation_precisions.append(validation_precision)

    validation_recall = recall_score(validation_trues, validation_predictions, average="weighted", zero_division=0)
    validation_recalls.append(validation_recall)

    print("\tDone!")
```
